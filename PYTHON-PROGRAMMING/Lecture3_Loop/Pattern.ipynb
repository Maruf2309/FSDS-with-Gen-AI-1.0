# Patterns Printing

✅ 1. Half Pyramid (Your example)

````python
row = int(input("Enter rows: "))

for i in range(1, row+1):
    for j in range(1, i+1):
        print("*", end="")
    print()
````

✔ Explanation

- Outer loop → controls rows
- Inner loop → prints stars equal to row number

✅ 2. Inverted Half Pyramid
```python
*****
****
***
**
*
```

````python
row = int(input("Enter rows: "))  # input 5

for i in range(row, 0, -1):
    for j in range(1, i+1):
        print("*", end="")
    print()
````

`range(5, 0, -1) → 5, 4, 3, 2, 1`

```python
So the loop runs 5 times with:

1st iteration → i = 5
2nd iteration → i = 4
3rd iteration → i = 3
4th iteration → i = 2
5th iteration → i = 1
```

This makes the pyramid decrease each row.

**✔ 3. Inner Loop → Prints stars**

```python
for j in range(1, i+1):
    print("*", end="")

When i = 5 → prints 5 stars
When i = 4 → prints 4 stars
When i = 3 → prints 3 stars
When i = 2 → prints 2 stars
When i = 1 → prints 1 star
```


✅ 5. Full Pyramid

```python
    *
   ***
  *****
 *******
*********
```

````python
rows = int(input("Rows: "))

for i in range(1, rows+1):
    print(" " * (rows - i), end="")
    print("*" * (2*i - 1))
````

✔ Explanation

- `" " * (rows - i)` → prints spaces
- `"*" * (2*i - 1)` → prints odd number of stars

```python

*   *
*****
```

````python
n = int(input("Size: "))

for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*", end="")
        else:
            print(" ", end="")
````
````python
    print()
````

⭐ 10. Pascal’s Triangle

```yaml
     1
    1 1
   1 2 1
  1 3 3 1
```

````python
rows = int(input("Rows: "))

for i in range(rows):
    print(" " * (rows - i), end="")
    num = 1
    for j in range(i+1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)  # nCr formula
    print()
````

⭐ 13. Butterfly Pattern

```python
*      *
**    **
***  ***
********
***  ***
**    **
*      *

```


````python
rows = int(input("Rows: "))

# Upper part
for i in range(1, rows+1):
    print("*" * i + " " * (2*(rows-i)) + "*" * i)

# Lower part
for i in range(rows-1, 0, -1):
    print("*" * i + " " * (2*(rows-i)) + "*" * i)
````

⭐ 14. Number Diamond

```python
   1
  121
 12321
1234321
 12321
  121
   1
```

````python
rows = int(input("Rows: "))

# Upper part
for i in range(1, rows+1):
    print(" " * (rows-i), end="")
    for j in range(1, i+1):
        print(j, end="")
    for j in range(i-1, 0, -1):
        print(j, end="")
    print()

# Lower part
for i in range(rows-1, 0, -1):
    print(" " * (rows-i), end="")
    for j in range(1, i+1):
        print(j, end="")
    for j in range(i-1, 0, -1):
        print(j, end="")
    print()
````

⭐ 15. Hollow Pyramid

```python
    *
   * *
  *   *
 *     *
*********
```

````python
rows = int(input("Rows: "))

for i in range(1, rows+1):
    print(" " * (rows - i), end="")
    
    for j in range(1, 2*i):
        if j == 1 or j == 2*i - 1 or i == rows:
            print("*", end="")
        else:
            print(" ", end="")
    print()
````

⭐ 16. Hourglass Pattern

```python
***********
 *********
  *******
   *****
    ***
     *
    ***
   *****
  *******
 *********
***********

```

````python
rows = int(input("Rows: "))

# Upper inverted pyramid
for i in range(rows, 0, -1):
    print(" " * (rows - i) + "*" * (2*i - 1))

# Lower normal pyramid
for i in range(2, rows+1):
    print(" " * (rows - i) + "*" * (2*i - 1))
````

⭐ 28. Snake Number Pattern
```python
1 2 3 4
8 7 6 5
9 10 11 12
16 15 14 13

```

````python
n = int(input("Size: "))
num = 1

for i in range(n):
    row = []
    for j in range(n):
        row.append(num)
        num += 1
    if i % 2 == 1:
        row.reverse()
    print(*row)
````

⭐ 31. Zigzag Pattern
```python
    *       *
  *   *   *   *
*       *       *

```

````python
n = int(input("Columns: "))

for i in range(3):
    for j in range(1, n+1):
        if (i + j) % 4 == 0 or (i == 1 and j % 4 == 0):
            print("*", end="")
        else:
            print(" ", end="")
    print()
````
