# ATM / Banking-like operation

```python
import logging


# Step 1: Configuration logging
logging.basicConfig(
    filename="bank_app.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

def withdrawal(balance, amount):
    logging.info(f"Withdrawal Requested {amount}")  

    if amount <=0:
        logging.error("Invalid withdrawal amount!")   
        return "Invalid Amount"

    if amount > 0: 
        logging.warning("Insufficient Balance")      
        return "Insufficient Balance"         


    balance -= amount
    logging.info(f"Withdrawal Successful. Remaining balance {balance}")   

# Step 2: Use the function
balance = 5000

withdrawal(balance, 1000)
withdrawal(balance, 7000)
withdrawal(balance, 800)
withdrawal(balance, 0)
```

🧠 Why logging (real-life benefit)

| `print()`          | `logging`              |
| ------------------ | ---------------------- |
| Temporary          | Permanent record       |
| No level           | INFO / WARNING / ERROR |
| Not for production | Production-ready       |
| Hard to debug      | Easy audit trail       |
