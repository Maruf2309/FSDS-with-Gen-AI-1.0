# Logging in Python

Logging in Python is a **professional way to record program activity** (info, warnings, errors) instead of using `print()`.

Below is a clear, practical explanation with exam + real-life examples.

🔹 What is Logging in Python?

**Logging** is used to **track events** that happen when a program runs, such as:
- Errors
- Warnings
- Program flow
- Debug Information

📌 Used in **real projects, data pipelines, ML systems, production apps**

🔹 Why use Logging instead of `print()?`


| `print()`                  | `logging`               |
| -------------------------- | ----------------------- |
| Temporary debugging        | Professional monitoring |
| No severity level          | Multiple levels         |
| Cannot save to file easily | File + console          |
| Not configurable           | Fully configurable      |


🔹 Logging Levels (VERY IMPORTANT – Exam)

| Level      | Use case                     |
| ---------- | ---------------------------- |
| `DEBUG`    | Detailed info for developers |
| `INFO`     | Program running normally     |
| `WARNING`  | Something unexpected         |
| `ERROR`    | Program failed               |
| `CRITICAL` | System crash                 |


🔹 Basic Logging Example

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.debug("This is a debug message")
logging.info("Program started")
logging.warning("This is a warning")
logging.error("An error occurred")
logging.critical("Critical system failure")
```

🔹 Logging to a File (Very Common)

```python
import logging

logging.basicConfig(
    filename="app.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Application started")
logging.error("File not found")
```

🔹 Logging with Exception Handling (Real-Life Example)

```python
import logging

logging.basicConfig(level=logging.INFO)

try:
    a = 10
    print(a / 0)
except Exception as e:
    logging.error("Error occurred", exc_info=True)
```

🔹 Real Data Analysis Example (Your Domain)

```python
import logging
import pandas as pd

logging.basicConfig(level=logging.INFO)

try:
    df = pd.read_csv("sales_data.csv")
    logging.info("Sales data loaded successfully")
except FileNotFoundError:
    logging.error("Sales data file not found")
```

## Chapter 2

🔹 Advanced Logging in Python (Professional Level)

1️⃣ Logging to Console + File (Best Practice)

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(name)s | %(message)s",
    handlers=[
        logging.FileHandler("app.log"),
        logging.StreamHandler()
    ]
)

logging.info("Application started")
logging.warning("This is a warning")
logging.error("An error occurred")
```

2️⃣ Rotating Log Files (VERY IMPORTANT – Interview Favorite)

Prevents log files from becoming too large.

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("my_app")
logger.setLevel(logging.INFO)

handler = RotatingFileHandler(
    "app.log",
    maxBytes=2_000_000,  # 2 MB
    backupCount=5
)

formatter = logging.Formatter(
    "%(asctime)s | %(levelname)s | %(message)s"
)
handler.setFormatter(formatter)

logger.addHandler(handler)

logger.info("This is a rotating log example")
```

Logging in a Data Science / ML Pipeline (Your Domain)

```python
import logging
import pandas as pd

logging.basicConfig(
    filename="pipeline.log",
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)

def load_data(path):
    try:
        df = pd.read_csv(path)
        logging.info(f"Loaded data from {path}, rows={len(df)}")
        return df
    except FileNotFoundError:
        logging.error(f"File not found: {path}")
        return None

df = load_data("sales_data.csv")
```
