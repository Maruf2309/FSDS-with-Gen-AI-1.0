# Real-Life Sales CSV Processing using Pandas + Logging

```python
import pandas as pd
import logging 


# 1️⃣ Logging configuration
logging.basicConfig(
    filename="sales_data.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
# 2️⃣ Function to read sales data
def read_sales_data(file_path):
    try:
        data = pd.read_csv(file_path)
        logging.info(f"Successfully read data from {file_path}")
        return data
    except Exception as e:
        logging.error(f"Error reading data from {file_path}: {e}")
        return None

```
