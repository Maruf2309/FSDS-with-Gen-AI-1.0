```python
import pandas as pd
import logging
import os

# -----------------------
# Logger setup
# -----------------------
logger = logging.getLogger("sales_logger")
logger.setLevel(logging.INFO)

# Formatter
formatter = logging.Formatter("%(asctime)s - %(levelname)s - %(message)s")

# Console Handler
ch = logging.StreamHandler()
ch.setFormatter(formatter)
logger.addHandler(ch)

# File Handler
fh = logging.FileHandler("sales_analysis2.log")
fh.setFormatter(formatter)
logger.addHandler(fh)

# -----------------------
# CSV file path
# -----------------------
file_path = "D:/Dataset/db_sales.csv"

# -----------------------
# Load CSV safely
# -----------------------
if not os.path.exists(file_path):
    logger.error(f"CSV file '{file_path}' not found!")
    raise FileNotFoundError(f"CSV file '{file_path}' not found!")

try:
    sales_df = pd.read_csv(file_path, encoding="utf-8")
    logger.info(f"Loaded CSV file successfully with {len(sales_df)} rows")
except UnicodeDecodeError:
    logger.warning("UTF-8 decoding failed, trying 'cp1252' encoding")
    sales_df = pd.read_csv(file_path, encoding="cp1252")
    logger.info(f"Loaded CSV file successfully with {len(sales_df)} rows using cp1252 encoding")

# -----------------------
# Data Cleaning / Checks
# -----------------------
# Check for negative or zero sales amounts
negative_sales = sales_df[sales_df['sales_amount'] <= 0]
if not negative_sales.empty:
    for idx, row in negative_sales.iterrows():
        logger.warning(f"Negative or zero sales found: Client '{row['client_name']}' - Amount: {row['sales_amount']}")

# -----------------------
# Aggregation / Analysis
# -----------------------
# Total sales per client
total_sales_client = sales_df.groupby('client_name')['sales_amount'].sum().reset_index()
total_sales_client.rename(columns={'sales_amount': 'Total_Sales'}, inplace=True)
logger.info("Aggregated total sales per client")
print(total_sales_client)

# High sales alert (example: sales > 1,000,000)
high_sales_threshold = 1_000_000
high_sales = sales_df[sales_df['sales_amount'] > high_sales_threshold]
if not high_sales.empty:
    for idx, row in high_sales.iterrows():
        logger.warning(f"High sales alert: Client '{row['client_name']}' - Amount: {row['sales_amount']} on {row['date']}")

# -----------------------
# Summary
# -----------------------
logger.info("Sales Analysis Completed!")
```
