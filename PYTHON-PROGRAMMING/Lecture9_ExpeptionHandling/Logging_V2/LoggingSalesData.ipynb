# Realistic example of logging in Python using sales data.

```python
import pandas as pd
import logging
import os

# -----------------------
# Clear existing logging handlers (Important for Jupyter / VS Code)
# -----------------------
for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)

# -----------------------
# Setup Logger
# -----------------------
logger = logging.getLogger("sales_logger")  # unique logger
logger.setLevel(logging.INFO)

# Console handler
ch = logging.StreamHandler()
ch.setLevel(logging.INFO)

# File handler (creates a log file)
fh = logging.FileHandler("sales_analysis.log")
fh.setLevel(logging.INFO)

# New-style {} formatting
formatter = logging.Formatter("{asctime} - {levelname} - {message}", style="{")
ch.setFormatter(formatter)
fh.setFormatter(formatter)

logger.addHandler(ch)
logger.addHandler(fh)

# -----------------------
# Example Sales Analysis
# -----------------------
logger.info("Sales Analysis Started")

# Sample sales data
data = {
    "Date": ["2026-01-03", "2026-01-04", "2026-01-05", "2026-01-06"],
    "Salesperson": ["Alice", "Bob", "Carol", "Alice"],
    "Units": [15, 5, 8, 10],
    "Unit_Price": [50, 30, 50, 50]
}
sales_df = pd.DataFrame(data)
logger.info(f"Loaded sales data successfully with {len(sales_df)} rows")

# Calculate Total Sale
sales_df['Total_Sale'] = sales_df['Units'] * sales_df['Unit_Price']
logger.info("Calculated Total_Sale column")

# Aggregate by Salesperson
sales_summary = sales_df.groupby('Salesperson')['Total_Sale'].sum().reset_index()
logger.info("Aggregated total sales by Salesperson")
print(sales_summary)

# Warn if units sold > 10
for _, row in sales_df.iterrows():
    if row['Units'] > 10:
        logger.warning(f"High units sold by {row['Salesperson']} on {row['Date']}: {row['Units']} units")

logger.info("Sales Analysis Completed!")
```
