# Exception Handling in Python

Exception handling is used to handle runtime errors so your program doesn’t crash and can respond gracefully.

An exception is an error that occurs during program execution which disrupts the normal flow of a program.

👉 If an exception is not handled, the program stops (crashes).
- Simple Term : `Exception = Runtime Error`

There are 2 stages where error may happen in a program

- During compilation -> Syntax Error
- During execution -> Exceptions


### Syntax Error

- Something in the program is not written according to the program grammar.
- Error is raised by the interpreter/compiler
- You can solve it by rectifying the program

Example:

`print "Hello World" --> Here parenthesis is missing, that why its Syntax Error`

### Execution Error
- Exception = Runtime error

### Other examples of syntax error

- Leaving symbols like colon,brackets
- Misspelling a keyword
- Incorrect indentation
- empty if/else/loops/class/functions

What is a Syntax Error?
- A Syntax Error occurs when Python cannot understand the structure (grammar) of your code.
- 📌 It is detected before execution (at compile time).

#### Common Types of Syntax Errors (with Examples)
- Missing Colon
- Unmatched Parentheses / Brackets
- Incorrect Indentation
- Missing Quotes
- Using Keywords Incorrectly
- Invalid Variable Name
- Extra or Missing Operators
- Unmatched Parentheses / Brackets
- Attribute Error
