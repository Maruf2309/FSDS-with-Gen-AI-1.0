# Exception Handling in Python

Exception handling is used to handle runtime errors so your program doesn’t crash and can respond gracefully.

An exception is an error that occurs during program execution which disrupts the normal flow of a program.

👉 If an exception is not handled, the program stops (crashes).
- Simple Term : `Exception = Runtime Error`

There are 2 stages where error may happen in a program

- During compilation -> Syntax Error
- During execution -> Exceptions


### Syntax Error

- Something in the program is not written according to the program grammar.
- Error is raised by the interpreter/compiler
- You can solve it by rectifying the program

Example:

`print "Hello World" --> Here parenthesis is missing, that why its Syntax Error`

### Execution Error
- Exception = Runtime error

````python
# Valid Case
def divide(num):
    r = 10 / num
    print(f"10 divided by {num} is : {r}")    

divide(2)  # Valid Case, but user may input any number or string, that come amout error handing
````

````python
# invalid Case
def divide(num):    
    r = 10 / num
    print(f"10 divided by {num} is : {r}")

divide(0)  # invalid Case, ZeroDivisionError

# The program is terminated here due to unhandled exception
# To handle such exceptions, we use try and except blocks
# The code which may raise exception is placed inside try block
````

````python
try:
    def divide(num):    
        r = 10 / num
        print(f"10 divided by {num} is : {r}")

    divide(0)  # invalid Case, ZeroDivisionError
except ZeroDivisionError:
    print("You cannot divide a number by zero")
# The program continues to run after handling the exception
print("Program continues after exception handling") 
````

````python
try:
    def divide(num):
        x= 10/ num
        print(f"10 is divided by {num} is :{x}")

    divide("a")  # invalid Case, TypeError

except TypeError:
    print("Invalid input type. Please provide a number.")
# The program continues to run after handling the exception
print("Program continues after exception handling")
````

````python
try:
  def divide(num):
    x = 10 / num
    print(f"10 divided by {num} is {x}")

  divide("a")

except Exception as e:
  print(e)
````

````python
def check_positive(num):
    if num < 0:
        raise ValueError("The number is negative. Please provide a positive number.")
    else:
        print(f"The number {num} is positive.") 
check_positive(-5)  # This will raise ValueError
````

````python
def divide(a,b):
    if b == 0:
        raise ZeroDivisionError("Denominator cannot be zero.")
    else:
        result = a / b
        print(f"The result of {a} divided by {b} is {result}")  
divide(10,0)  # This will raise ZeroDivisionError
````
