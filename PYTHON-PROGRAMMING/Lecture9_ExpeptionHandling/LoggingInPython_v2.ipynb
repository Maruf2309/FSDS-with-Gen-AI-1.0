# Logging in Python- V2

```python
import logging # this is a built-in module to handle logging in python
import os # its help to create directory (copy, delete, move files and directories)
from datetime import datetime
from logging import handlers
```

```python
LOG_DIR = 'logs'  # Create a directory named 'logs' to store log files
LOG_FILE_NAME = 'application.log'    # Log file name


# makedirs function will create directory if not present
os.makedirs(LOG_DIR, exist_ok=True) # exist_ok=True will not raise error if directory is already present (not allow to re-create same directory)

# path definition for log file(Manual Approach)
# path = "logs/application.log" # Manual Approach

# Programmatic Approach
log_path = os.path.join(LOG_DIR, LOG_FILE_NAME)
print(log_path)


# logging configuration
logging.basicConfig(
    filename=log_path,
    level=logging.INFO,   # logging level (DEBUG, INFO, WARNING, ERROR, CRITICAL)
    format='[%(asctime)s] %(levelname)s - %(message)s',  # log message format
    datefmt='%Y-%m-%d %H:%M:%S'  # date format
)

```

```python
# Application code
def profile(name):
    if type(name) == str:
        print(f"My name is {name}")
    else:
        logging.info("Your input is not valid")

# calling function
profile("John Doe")
profile(12345)  # This will log an info message since the input is not a
```

```python
# Example of exception handling with logging
def divide_numbers(a, b):
    try:
        result = a / b
        logging.info(f"Division successful: {a} / {b} = {result}")
        return result
    except ZeroDivisionError as e:
        logging.error("Attempted to divide by zero.")
        logging.exception("Exception occurred")  # This will log the stack trace
    except Exception as e:
        logging.error("An unexpected error occurred.")
        logging.exception("Exception occurred")  # This will log the stack trace

# calling divide_numbers function
divide_numbers(10, 2)
divide_numbers(10, 0)  # This will log an error for division by zero
divide_numbers(10, 'a')  # This will log an unexpected error
```

```python
def devide_numbers(a, b):
    if b == 0:
        logging.info("Division by zero is not allowed.")
    else:
        result = a / b
        logging.info(f"Division successful: {a} / {b} = {result}")
        return result
```
