# Logging in Python- V2

```python
import logging
import os # its help to create directory (copy, delete, move files and directories)
from datetime import datetime
from logging import handlers
```

```python
LOG_DIR = 'logs'  # Create a directory named 'logs' to store log files
LOG_FILE_NAME = 'application.log'    # Log file name


# makedirs function will create directory if not present
os.makedirs(LOG_DIR, exist_ok=True) # exist_ok=True will not raise error if directory is already present (not allow to re-create same directory)

# path definition for log file(Manual Approach)
# path = "logs/application.log" # Manual Approach

# Programmatic Approach
log_path = os.path.join(LOG_DIR, LOG_FILE_NAME)
print(log_path)
```
