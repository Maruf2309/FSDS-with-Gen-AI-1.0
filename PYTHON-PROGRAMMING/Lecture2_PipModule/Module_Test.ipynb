# Module Exercise

## 📦 1. math — Mathematical Superpower

```python
# "Use the math module when you need precise mathematical functions."

import math
print(math.sqrt(25))
print(math.pi)
```

### 🎲 2. random — Random Number Generator

```python
# "Perfect for games, simulations, and AI experiments."
import random
print(random.randint(1,5))
```

###  📅 3. datetime — Work With Dates & Time

```python
# "Used in almost every real-world application — logs, timestamps, schedules."
from datetime import datetime
print(datetime.now())
```

### 🔑 4. keyword — Python Reserved Words

```python
# "Use this when you want to check if a word is a Python keyword."
import keyword
print(keyword.kwlist)
```

### 🔤 5. string — Useful String Constants

```python
# "Great for validation, password generation, and text processing."
import string
print(string.ascii_letters)
```

### 📁 6. os — Work With Files & Folders

```python
# "Used to create folders, get paths, and read system info."
import os
print(os.getcwd())
```

### 🖥️ 7. sys — System & Interpreter Info

```python
import sys
print(sys.version) # Python Version
```

### 🧪 8. re — Regular Expressions

```python
# "Search, extract, and validate text."
import re
print(re.findall(r"\d+", "AB55CD99"))
```

### 🎯 9. json — Work With JSON Data

```python
# "Essential for APIs, data exchange, and config files."
import json
data = json.loads('{"name": "Maruf"}')
print(data["name"])
```

### 📊 10. statistics — Mean, Median, Standard Deviation

```python
import statistics as st
print(st.mean([10, 20, 30]))
```
