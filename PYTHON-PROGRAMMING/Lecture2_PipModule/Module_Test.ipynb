```python
# Use random.choice() when you want one name.
import random 
names = ["Maruf", "Rakib", "Siam", "Hasan", "Tamim"]

random_name = random.choice(names)
print(random_name)
```

```python
# Select Multiple Random Names (with duplicates allowed)
import random
names = ["Maruf", "Rakib", "Siam", "Hasan", "Tamim"]
selected = random.choices(names, k=3) #  # select 3 names
print(selected)
```

```python
# Select Multiple Names WITHOUT duplicates
## Use random.sample() → No repetition.

import random

names = ["Maruf", "Rakib", "Siam", "Hasan", "Tamim"]

selected = random.sample(names, 3)  # pick 3 unique names
print(selected)
```

```python
# 🎯 1. Random Winner Selector (like lottery / giveaway)

import random

names = ["Maruf", "Rakib", "Siam", "Hasan", "Tamim","Reza","Mitu","Tasneem"]

winner = random.choice(names)
print("Winner:", winner)
```

```python
# Pick 3 Random Winners (no duplicates)
import random

names = ["Maruf", "Rakib", "Siam", "Hasan", "Tamim", "Jahid", "Naeem"]

winners = random.sample(names, 3)   # pick 3 unique winners
print("Selected Winners:", winners)
```

```python
# Random Team Generator (2 teams)

import random

names = ["Maruf", "Rakib", "Siam", "Hasan", "Tamim", "Jahid", "Naeem", "Kadir"]

random.shuffle(names)

team1 = names[:4]
team2 = names[4:]

print("Team 1:", team1)
print("Team 2:", team2)
```

```python
# Random Group Assignment (Team A, Team B, Team C)

import random

names = ["Maruf", "Rakib", "Siam", "Hasan", "Tamim", "Jahid", "Naeem", "Kadir", "Shuvo"]

groups = {"A": [], "B": [], "C": []}

for name in names:
    group = random.choice(list(groups.keys()))
    groups[group].append(name)

print(groups)
```

```python
# Random Name for Daily Task (Task Rotation)

import random

names = ["Maruf", "Rakib", "Siam", "Hasan"]

today_person = random.choice(names)
print("Today's responsible person:", today_person)
```

```python
# Random Name Password (fun use)
import random
import string

names = ["Maruf", "Rakib", "Siam", "Hasan"]

random_name = random.choice(names)
password = random_name + ''.join(random.choices(string.digits, k=3))

print("Generated Password:", password)
```

```python
# Random Seating Arrangement

import random

students = ["Maruf", "Rakib", "Siam", "Hasan", "Tamim", "Naeem"]

random.shuffle(students)
print("New seating:", students)
```

```python
# Random Group of Size N (User Input)

import random

names = ["Maruf", "Rakib", "Siam", "Hasan", "Tamim", "Naeem"]

n = 2   # group size
group = random.sample(names, n)
print("Selected Group:", group)
```
