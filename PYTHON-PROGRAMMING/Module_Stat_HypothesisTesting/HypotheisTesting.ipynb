# Hypothesis Testing, probability density, Mass & Cumulative Density Function

Hypothesis testing is a statistical method to make decisions using data instead of assumptions.

🔴 Real-Life Scenario (Very Practical)

🏭 Business Case: Sales Improvement Check

- You work in an electrical goods company (fans, cables, lights).

👉 Marketing claims:

- “After the new promotion, average monthly sales increased.”
- As a data person, you ask:
- Is this real improvement or just random variation?
- That’s where hypothesis testing comes in.

````python
sales_before = [52, 55, 50, 48, 53, 54, 49, 51]
sales_after  = [58, 60, 57, 59, 61, 56, 60, 62]
````

4️⃣ Perform Hypothesis Test (Python)

````python
from scipy.stats import ttest_rel

t_stat, p_value = ttest_rel(sales_after, sales_before)

print("t-statistic:", t_stat)
print("p-value:", p_value)
````

5️⃣ Decision Rule

Significance Level (α):

````python
α = 0.05
````

| Condition | Decision          |
| --------- | ----------------- |
| p ≤ 0.05  | Reject H₀         |
| p > 0.05  | Fail to reject H₀ |


6️⃣ Interpretation (MOST IMPORTANT)

````python
p_value = 0.002
````

✔ Since 0.002 < 0.05

✅ Final Conclusion:

We reject the null hypothesis.

There is strong statistical evidence that the promotion increased sales.

📌 This is exactly how management decisions are justified.

## Chapter 2: 📊 Full Python Code: Hypothesis Testing (Real-Life Example)

1️⃣ Import Required Libraries

````python
import numpy as np
import pandas as pd
from scipy.stats import ttest_rel
````

2️⃣ Realistic Sales Data (Sample)

(Monthly sales in lakh BDT for same branches)

````python
sales_before = np.array([52, 55, 50, 48, 53, 54, 49, 51])
sales_after  = np.array([58, 60, 57, 59, 61, 56, 60, 62])
````

3️⃣ Create a DataFrame (Good Practice)

````python
df = pd.DataFrame({
    'Sales_Before': sales_before,
    'Sales_After': sales_after
})
print(df)
````

4️⃣ Define Hypotheses

````python
"""
H0 (Null Hypothesis): Mean sales after promotion <= mean sales before promotion
H1 (Alternative Hypothesis): Mean sales after promotion > mean sales before promotion
"""
````

5️⃣ Perform Paired T-Test

````python
t_stat, p_value = ttest_rel(sales_after, sales_before)

print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")
````

6️⃣ Decision Logic (Very Important)

````python
alpha = 0.05  # significance level

if p_value < alpha:
    print("✅ Reject Null Hypothesis")
    print("📈 Promotion significantly increased sales.")
else:
    print("❌ Fail to Reject Null Hypothesis")
    print("📉 No significant increase in sales.")
````

7️⃣ Business-Friendly Summary (Best Practice)

````python
mean_before = sales_before.mean()
mean_after = sales_after.mean()

print("\n--- Business Summary ---")
print(f"Average Sales Before Promotion: {mean_before:.2f} lakh BDT")
print(f"Average Sales After Promotion : {mean_after:.2f} lakh BDT")
print(f"Increase in Average Sales     : {mean_after - mean_before:.2f} lakh BDT")
````

🧠 Expected Interpretation (Human Language)

- Since the p-value is less than 0.05, the increase in sales is statistically significant.
- The promotion successfully improved average monthly sales.

📌 Why This Example Is REALISTIC

- ✔ Same branches → paired test
- ✔ Numeric KPI (sales)
- ✔ Clear business impact
- ✔ Director-level explanation possible

🎓 Interview-Ready Explanation (One Paragraph)

```bash
We conducted a paired t-test to compare average monthly sales before and after a marketing promotion. The p-value was below the 5% significance level, indicating that the observed increase in sales is statistically significant and unlikely due to random variation.
```

# 📊 Hypothesis Testing WITH Visualization (Python – Realistic)

Step 0: Why visualization first?

Before any hypothesis test, always visualize:

- Detect skewness
- Spot outliers
- Check assumption sanity

1️⃣ Import Libraries

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import ttest_rel
````

2️⃣ Realistic Dataset (Sales Before vs After)

````python
data = {
    "Sales_Before": [5200000, 5500000, 5000000, 4800000, 5300000, 5400000, 4900000, 5100000],
    "Sales_After":  [5800000, 6000000, 5700000, 5900000, 6100000, 5600000, 6000000, 6200000]
}

df = pd.DataFrame(data)
df
````

3️⃣ Visualization (Histogram + KDE)

````python
plt.figure(figsize=(10, 6))

plt.subplot(1, 2, 1)
sns.histplot(df['Sales_Before'],kde=True, color='blue', bins=10)
plt.title('Sales Before Promotion')

plt.subplot(1, 2, 2)
sns.histplot(df['Sales_After'], kde=True, color='green', bins=10)
plt.title('Sales After Promotion')

plt.tight_layout()
plt.show()
````

🔍 Insight you look for:

- Is after-sales distribution shifted right?
- Any extreme outliers?

4️⃣ Boxplot (Outlier & Median Comparison)

````python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, palette=['blue', 'green'])
plt.title('Boxplot of Sales Before and After Promotion')
plt.ylabel('Sales Amount (in BDT)') 
plt.show()
````

📌 This instantly shows:
- Median increase (that sales sales increased)
- The median and overall sales distribution shifted upward after the promotion.
- Spread change
- Outliers (if any)

5️⃣ Define Hypotheses (Business Language)

```bash
H0: Average sales after promotion ≤ average sales before promotion
H1: Average sales after promotion > average sales before promotion
```

6️⃣ Perform Paired T-Test

````python
t_stat, p_value = ttest_rel(df['Sales_After'], df['Sales_Before'])
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")
````

7️⃣ Decision Logic (Industry Standard)

````python
# Hypothesis Testing Decision (case1)

alpha = 0.05  # significance level

if p_value < alpha:
    print("✅ Reject Null Hypothesis")
    print("📈 Promotion significantly increased sales.")
else:
    print("❌ Fail to Reject Null Hypothesis")
    print("📉 No significant increase in sales.")

print("\n--- Business Summary ---")
mean_before = df['Sales_Before'].mean()
mean_after = df['Sales_After'].mean()
print(f"Average Sales Before Promotion: {mean_before:,.2f} BDT")
print(f"Average Sales After Promotion : {mean_after:,.2f} BDT")
print(f"Increase in Average Sales     : {mean_after - mean_before:,.2f} BDT")

````

````python
# Hypothesis Testing Decision (case2)/ same as case1
alpha = 0.05

if p_value < alpha:
    decision = "Reject H0"
    conclusion = "Promotion significantly increased sales"
else:
    decision = "Fail to Reject H0"
    conclusion = "No significant sales improvement"

print("\nDecision:", decision)
print("Conclusion:", conclusion)
````

- We applied a standard statistical test used globally to verify whether this increase is real or random
- There is less than a 5% chance that this sales increase happened by luck.
- “This confirms that the promotion strategy worked and can be safely scaled to other regions or product lines.”
- The data shows consistent improvement across branches, not dependency on a few outliers, this builds trust in the data

8️⃣ Business Summary (VERY IMPORTANT)

````python
print("\n--- Business Summary ---")
print(f"Average Sales Before: {df['Sales_Before'].mean():,.0f}")
print(f"Average Sales After : {df['Sales_After'].mean():,.0f}")
print(f"Increase in Sales   : {(df['Sales_After'].mean() - df['Sales_Before'].mean()):,.0f}")
````

🧠 Final Interpretation (Manager-Friendly)

```bash
The statistical test confirms that the increase in sales after the promotion is significant at a 5% confidence level. This improvement is not due to random variation and reflects a real positive business impact.
```
