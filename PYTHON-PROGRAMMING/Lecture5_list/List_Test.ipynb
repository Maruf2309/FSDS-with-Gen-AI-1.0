Exercise 6: Check if list is palindrome

````python
L = [1,2,3,2,1]

if L == L[::-1]:
    print("Palindrome")
````

**What is a palindrome?**

A palindrome is something that reads the same forward and backward.

`Examples: 121, madam, [1,2,3,2,1]`.

🔍 How the code works
1. lst[::-1] reverses the list.

Python slicing with `[::-1]` means:

- Start from the end
- Go backward step by step
- Result → reversed list

Compare original list with reversed list

```bash
lst == lst[::-1]

[::-1]

[start : stop : step]

start → empty → start from the end
stop → empty → go all the way to the beginning
step = -1 → move backward one step at a time


```

````python
# List Palindrome (like your example)

lst = [4, 5, 6, 5, 4]

if lst == lst[::-1]:
    print("List is a Palindrome")
else:
    print("List is NOT a Palindrome")
````

````python
# String Palindrom

s = "madam"

if s == s[::-1]:
    print("String is a Palindrome")
else:
    print("String is NOT a Palindrome")
````

````python
# Number Palindrome

num = 12321

# Convert number to string to reverse
if str(num) == str(num)[::-1]:
    print("Number is a Palindrome")
else:
    print("Number is NOT a Palindrome")
````

````python
s = "racecar"
is_palindrome = True

for i in range(len(s)//2):
    if s[i] != s[-(i+1)]:
        is_palindrome = False
        break

if is_palindrome:
    print("Palindrome")
else:
    print("Not a Palindrome")
````

Exercise 7: Flatten a nested list

````python
nested = [[1, 2], [3, 4], [5]]

flat = [item for sub in nested for item in sub]

print(flat)
````

````python
# It is equivalent to:

flat = []
for sub in nested:     # outer loop → each inner list
    for item in sub:   # inner loop → each element inside inner list
        flat.append(item)
print(flat)
````

Exercise 8: Find frequency of each element

````python
lst = [1, 2, 2, 3, 3, 3]
freq = {}

for i in lst:
    freq[i] = freq.get(i, 0) + 1

print(freq)
````

````python
# 🔥 Find the Most Frequent Element
from collections import Counter

lst = [1, 2, 2, 3, 3, 3]

freq = Counter(lst)
print(freq.most_common(1))
````

````python
# Find All Elements With Max Frequency
lst = [1, 2, 2, 3, 3, 3]

freq = {}

for i in lst:
    freq[i] = freq.get(i, 0) + 1

max_freq = max(freq.values())
most_frequent_items = [k for k, v in freq.items() if v == max_freq]

print("Max Frequency:", max_freq)
print("Items:", most_frequent_items)

````

````python
# Duplicate Elements Only
lst = [1,2,2,3,4,4,4,5]

freq = {}
for i in lst:
    freq[i] = freq.get(i, 0) + 1

for k, v in freq.items():
    if v > 1:
        print(k, "→", v, "times")
````

````python
# Most Frequent Element(s)
lst = [3,3,2,1,2,3,4,4]

freq = {}
for i in lst:
    freq[i] = freq.get(i, 0) + 1  # stores the updated count back into the dictionary by `freq[i]`

max_freq = max(freq.values())

most_frequent = [k for k, v in freq.items() if v == max_freq]

print("Most frequent:", most_frequent)
````

````python
# Character With Maximum Frequency
text = "success"

freq = {}  # empty dictionary for storing Character freq
for ch in text:
    freq[ch] = freq.get(ch, 0) + 1

max_freq = max(freq.values())

for k, v in freq.items():
    if v == max_freq:
        print(k, "→", v, "times")
````

- `freq.get(ch, 0) →` if the character exists, return its count, otherwise return 0
- + 1 → add 1 to count
- Store back in dictionary

| ch | freq before         | freq after                       |
| -- | ------------------- | -------------------------------- |
| s  | {}                  | {'s': 1}                         |
| u  | {'s': 1}            | {'s': 1, 'u': 1}                 |
| c  | {'s': 1, 'u': 1}    | {'s': 1, 'u': 1, 'c': 1}         |
| c  | count was 1         | count becomes 2                  |
| e  | add 'e':1           | {'s': 1, 'u': 1, 'c': 2, 'e': 1} |
| s  | s was 1 → becomes 2 |                                  |
| s  | s was 2 → becomes 3 |                                  |


Loop through dictionary:
```python
's' → 3 → matches max → print
'u' → 1 → ignore
'c' → 2 → ignore
'e' → 1 → ignore
```

````python
# Frequency in Nested List

nested = [[1,2],[2,3],[3,3,3]]

flat = []
for sub in nested:
    for item in sub:
        flat.append(item)

freq = {}
for i in flat:
    freq[i] = freq.get(i, 0) + 1

print(freq)

````

8. Rotate list by k positions

````python
def rotate(lst, k):
    k = k % len(lst)
    return lst[-k:] + lst[:-k]
````
