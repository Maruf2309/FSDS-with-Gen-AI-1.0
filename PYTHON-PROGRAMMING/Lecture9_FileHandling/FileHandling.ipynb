# Module 1 - Basic

**1.1 Writing to a file**

```python
# if the file is not present 

f = open("files/sample.txt", "w")
f.write("Hello World") # writing text inside the file created
f.close()

"""
1. Initially an empty folder name as "files" created
2. A file (sample.txt) created inside files by giving forward slash
3. Write "Hello World inside sample.txt
"""
```

**How it works**

- `open`: its a python function & created an object as `f`
- `w`: Define as write mode or allow to write somethin inside file, without `w` it would return error or yon cannot write
- `write`: this is a fuction to write something in file
- `close`: finally close the file


- `w`: it gives writing authority


**Steps**
- Open
- Write
- Close

**3.0 Write multiline strings**

```python
f = open('files/sample3.txt','w')
f.write("Hello World")          # Line 1
f.write('\nHow are you?')       # Line 2
f.write('Who are you ?- (Line3)')        # Line 3 (in same line)
f.write('\nI am fine')          # Line 4 (New Line)
f.close()
```

3.1 - Case 2 : If the file is already present

```python
f = open('files/sample4.txt','w')
f.write("Text Line 4")
f.close()
```

```python
# Now Text 4 will be replaced / overwrite by new line as `Text Line 5`
f = open('files/sample4.txt','w')
f.write("Text Line 5")   # Now previous line (4) replaced by text line 5
f.close()

# File will be repeatedly overwrites if you open it by `w` mode
```

**4.0 Append Mode**

- Suppose, you want to keep previous context & add new context into it, then use `Append` Mode denoted as `a`

```python
# Introducing append mode
f = open('files/sample.txt', 'a')
f.write('\nI am fine')  # now previous text will not be replaced, rather new text will added newly
f.close()
```

- Without `append` mode previous text will be replaced
- To keep previous text need to use `a` mode

**5.0 Write Lines**

- It expect a list of lines

```python
# Write lines
L = ['Hello\n','hi\n','how are you?\n', 'I am fine']  # list of lines/ sentence

f = open('files/sample5.txt','w')
f.writelines(L)  # expect list of lines
f.close()
```

**6.0 Reading from Files**

- Reading file by `python`
- Authorizing someone to `read` only

```python
f = open('files/sample5.txt','r')

s = f.read()  # without giving any param inside parenthesis, it will read all
print(s)
f.close()
```

```python
# You can limit to read character by passing limit inside param

f = open('files/sample.txt','r')
s = f.read(17) # Read 12 chatercter, specify limit
print(s)
f.close()
```

**7.0 ReadLines**

```python
# readlines() -> to read line by line

f = open('files/sample.txt','r')
print(f.readline(), end='')  # Line by line1
f.close()
```

```python
f = open('files/sample.txt','r')
print(f.readline(), end='')  # Line by line- 1
print(f.readline(), end='')  # Line by line- 2
f.close()
```

```python
# Read all lines by readlines with loop

f = open('files/sample.txt','r')

while True:
    data = f.readline() # read by readline infinite

## Break if empty str found (end of line)
    if data == "":
        break

    else:
        print(data, end="")

f.close()


#=======================================
# All can read with readline with loop, but its recommended to use `read`
#========================================
```
