# File Handling using Context Manager (WITH)

## Using Context Manager(with)

**This is Recommeded**

- Its good idea to close a file after usage as it free up the resources
- If we dont close it, garbase collector would close it
- `with` keyword closes the file as soon as the usages it over
- Regardless `close` the file, it automatically close the file
- `with open` generally need to save as object like end `as f`

```python
with open("files/sample.txt","w") as f:
    f.write("Hello there!")

# When open use, then its store as object 'f' no need to store as var, rather use as `f` will be store in f
# file not closed, but it automatically closed due to using `with open`
# `with` keyword it takes care automatically, or close forefully
# Sometime we can forget to close file, it close automatically
```

```python
# can not possible to add more due to using with
f.write("Maruf")  # return error, since it closed automatically due to using with
```

**1.0 Read File with Open**

```python
# Read()

with open("files/sample.txt", 'r') as f:
    s = f.read()
    print(s)

# File Closing is not mendatory anymore when used with open
```

```python
# Read file with limit
with open("files/sample.txt", "r") as f:
    print(f.read(10))
```

```python
# Readline()

with open("files/sample3.txt","r") as f:
    s = f.readline()
    print(s)

# Read line read as line
```

**3.0 Setting Cursor / Cursor Selection**

- Can set by giving `seek` parameter
- it works how we select `cursor` manually, just do this programitically

```python
with open("files/sample.txt","r") as f:
    f.seek(5)            # setting cursor, it will start reading from 5th Index position
    print(f.read())
```

`Tell()` tells the `cursor` present position

```python
# tell()
with open("files/sample.txt" ,"r") as f:
    f.seek(5)
    print(f.tell()) # Show the curson current position, as we select `5th index` earlier, so `there` is in 5th position
```

**Problems with working in text mode**

- Cannot work with other binary files like images
- not good for other data types like int/float/list/tuples

`Executable File`: Image (inside pixel value can not show)

```python
# Working with Binary file

with open("files/bird.jpg",'r') as f:
    print(f.read())

# it will give error, its a binary file so by only `r` we cannot reed it
# That why need 'rb`
```

```python
# Mode  `rb`
with open("files/bird.jpg",'rb') as f:   # mode `rb`
    print(f.read())

# We can show the output, but its not readable, since its binary data 
```

**2.2 Simple Operation(copy) over binary file**

- `read binary`, `write binary`

```python
# Need to copy so, write binary `rb`

with open("files/bird.jpg",'rb') as f:                  # read binary 'rb`
    original_file = f.read()                            # store in a variable

    with open("files/bird_copy.jpg",'wb') as wf:        # write binary (wf), we can write binary
        wf.write(original_file)
```

**Notes**

- Image file cannot read by only basic `python`- we need external library to process it

**What is JSON**

**JSON** stands for **JavaScript Object Notation**
- JSON is a standard text format for structuring and exchanging data between systems.

It is a lightweight, text-based data format used to store and exchange data between systems (for example, between a server and a web app).

**Why JSON is used**

- Easy for humans to read and write
- Easy for machines to parse and generate
- Language-independent (works with Python, Java, JavaScript, etc.)
- Very common in APIs, web apps, data engineering, and AI pipelines

**Basic JSON structure**

JSON is built on key–value pairs and arrays.

```python
{
    "name":"Maruf",
    "age":32,
    "skills":['Data Analysis', "AI","Machine Learning"],
}
```

**Key rules**

- Data is in key : value pairs
- Keys must be in double quotes
- Supported value types:
  - String
  - Number
  - Boolean (true / false)
  - Array [ ]
  - Object { }
  - null

**JSON vs Python Dictionary**
 - They look similar but are not the same.

 | JSON               | Python                  |
| ------------------ | ----------------------- |
| `true`             | `True`                  |
| `false`            | `False`                 |
| `null`             | `None`                  |
| Only double quotes | Single or double quotes |


**Where JSON is commonly used**

 - APIs (REST APIs send/receive JSON)

 - Web applications

 - Configuration files

 - Data storage & data exchange

 - Machine Learning / AI pipelines

 - NoSQL databases (MongoDB, etc.)

Example: JSON in API response

```python
{
  "status": "success",
  "data": {
    "user_id": 101,
    "email": "user@example.com"
  }
}
```

## Pickle

In python, **Pickle** is a built-in module used for **object serialization and deserialization**

Simply put:

`Picke`: Converting a Python object into a byte stream (save it) and loading it back later

🔹 Why use Pickle?

You use pickle when you want to:

- Save Python objects (list, dict, DataFrame, ML model, etc.)
- Reload them exactly as they were
- Avoid recomputing or retraining models

🔹 Basic Concept

| Action         | Meaning                      |
| -------------- | ---------------------------- |
| **Pickling**   | Save Python object to file   |
| **Unpickling** | Load Python object from file |


🔹 Example 1: Pickle a Python object (Save)

```python
import pickle

data = {
    "name": "Maruf",
    "role": "Data Analyst",
    "skills": ["Python", "SQL", "AI"]
}

# Save object
with open("data.pkl", "wb") as file:
    pickle.dump(data, file)

# 📁 File created: data.pkl
```

🔹 Example 2: Unpickle (Load)

```python
import pickle

with open("data.pkl", 'rb') as file:
    loaded_data = pickle.load(file)

print(loaded_data)
```

```python
import pickle

with open("data.pkl", 'rb') as file:
    loaded_data = pickle.load(file)

print(loaded_data)
print(type(loaded_data))
```

🔹 Example 3: Pickle a Machine Learning Model

```python
from sklearn.linear_model import LinearRegression
import pickle

# Experimental training data
X_train = [[2], [4], [6], [9]]   # must be 2D: list of lists
y_train = [4, 8, 12, 9]          # 1D list

# Create & train model
model = LinearRegression()
model.fit(X_train, y_train)

# Save model
with open("model.pkl",'wb') as f:
    pickle.dump(model, f)

print("Model trained and saved successfully.")
```

Load later

```python
with open("model.pkl","rb") as f:
    model = pickle.load(f)

predictions = model.predict(X_train)
predictions
```
